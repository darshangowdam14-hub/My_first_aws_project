```python
# Parking slots for each floor
floor1 = 5
floor2 = 5
floor3 = 5

while True:
    vehicle = input("Enter vehicle type (Car/Bike/Truck/exit): ").lower()

    if vehicle == "exit":
        break

    if vehicle != "bike" and vehicle != "car" and vehicle != "truck":
        print("Invalid Vehicle Type")
    else:
        hours = int(input("Enter parking hours: "))

        # Base price calculation
        if vehicle == "bike":
            rate = 20
        elif vehicle == "car":
            rate = 50
        else:
            rate = 100

        # Floor allocation
        assigned_floor = 0

        if vehicle == "truck":
            if floor1 > 0:
                assigned_floor = 1
                floor1 -= 1
            else:
                print("Parking Full")

        elif vehicle == "car":
            if floor1 > 0:
                assigned_floor = 1
                floor1 -= 1
            else:
                if floor2 > 0:
                    assigned_floor = 2
                    floor2 -= 1
                else:
                    print("Parking Full")

        else:  # bike
            if floor1 > 0:
                assigned_floor = 1
                floor1 -= 1
            else:
                if floor2 > 0:
                    assigned_floor = 2
                    floor2 -= 1
                else:
                    if floor3 > 0:
                        assigned_floor = 3
                        floor3 -= 1
                    else:
                        print("Parking Full")

        if assigned_floor != 0:
            # Base amount
            base_amount = rate * hours

            # Extra charges
            if hours <= 2:
                extra = 0
            elif hours <= 5:
                extra = base_amount * 0.10
            elif hours <= 10:
                extra = base_amount * 0.20
            else:
                extra = base_amount * 0.30

            total = base_amount + extra

            # Discount
            if vehicle == "bike":
                discount = total * 0.05
            elif vehicle == "car":
                discount = total * 0.10
            else:
                discount = total * 0.15

            final_bill = total - discount

            # Output
            print("\nVehicle Type :", vehicle.capitalize())
            print("Assigned Floor :", assigned_floor)
            print("Parking Hours :", hours)
            print("Base Amount :", base_amount)
            print("Extra Charges :", extra)
            print("Discount :", discount)
            print("Final Bill :", final_bill)

            print("\nRemaining Slots -")
            print("Floor 1 :", floor1)
            print("Floor 2 :", floor2)
            print("Floor 3 :", floor3)
            print()
```
