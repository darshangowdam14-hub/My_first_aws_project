```python
#Remove duplicates while preserving order

num = [1,2,2,3,4,4,5,2]
seen = set()
rev = []

for i in num:
    if i not in seen:
        seen.add(i)
        rev.append(i)

print(rev)        
```

```python
#Find duplicate IDs

num = [2,2,3,5,5,6,8,9]
seen = set()
res = []

for i in num:
    if i in seen not in res:
        res.append(i)
    seen.add(i)

print(res)        
```

```python
#Find missing number

num = [1,2,3,4,5,6,7,8,10]
n=0
for i in num:
    if i>n:
        n = i
        
total_sum = n * (n+1)//2
actual = sum(num)
missing = total_sum-actual
print(missing)
```

```python
# Find greatest with nested if
num = [100,49,68]
a,b,c = num
if a > b:
    great_num = a
    if a > c:
        great_num = a
    else:
        great_num = c    
else:
    great_num = b
    if c > b:
        great_num = c
print(great_num)    
```
